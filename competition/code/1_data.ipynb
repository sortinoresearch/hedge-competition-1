```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
# Download 5 years of daily data for the full competition universe
sector_universe = {
    "Technology / Semiconductor": {
        "equities": ["NVDA", "AMD", "INTC", "AVGO", "QCOM", "TSM", "ASML", "MU"],
        "etf": "SOXX",
    },
    "Financial / Banking": {
        "equities": ["JPM", "BAC", "MS", "GS", "C", "WFC", "PNC", "USB"],
        "etf": "XLF",
    },
    "Energy / Fossil & Clean": {
        "equities": ["XOM", "CVX", "COP", "SLB", "EOG", "MPC", "VLO", "DVN"],
        "etf": "XLE",
    },
    "Healthcare / Biotech": {
        "equities": ["LLY", "UNH", "JNJ", "ABBV", "MRK", "AMGN", "PFE", "BMY"],
        "etf": "XLV",
    },
}

equity_tickers = [
    ticker
    for sector in sector_universe.values()
    for ticker in sector["equities"]
]
etf_tickers = [sector["etf"] for sector in sector_universe.values()]
all_tickers = list(dict.fromkeys(equity_tickers + etf_tickers))

daily_data = yf.download(
    tickers=all_tickers,
    period="10y",
    interval="1d",
    auto_adjust=False,
    progress=False,
    threads=False,
)
daily_data = daily_data.dropna(how="all").sort_index()

if isinstance(daily_data.columns, pd.MultiIndex):
    downloaded_tickers = sorted(set(daily_data.columns.get_level_values(1)))
else:
    downloaded_tickers = all_tickers

missing_tickers = sorted(set(all_tickers) - set(downloaded_tickers))
print(f"Downloaded {len(downloaded_tickers)} tickers over {len(daily_data)} daily rows.")
print("Missing tickers:", missing_tickers if missing_tickers else "None")
daily_data.head()
```

```python
daily_data.to_csv("../data/daily_data.csv", index=True, header=True)
```
